```python
import random

def to_hex_str(value, width):
    return format(value, f'0{width}X')

input_filename = "pattern_input.data"
output_filename = "expected_output.data"

ac = 0 

with open(input_filename, "w") as fin, open(output_filename, "w") as fout:
    for _ in range(1000):
        x = random.getrandbits(16)  # Generate 16-bit random number
        ac = (ac + x) & 0xFFFFFFFF  

        # wirte patterns into input file
        fin.write(f"{to_hex_str(x, 4)}\n")
        # wirte patterns into output file
        fout.write(f"{to_hex_str(ac, 8)}\n")
```
